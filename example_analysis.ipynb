## 1. Import Libraries and Setup

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Enable inline plotting
%matplotlib inline

# Set visualization style
sns.set_theme(style="whitegrid", font_scale=1.1)
plt.rcParams['figure.figsize'] = (12, 8)

# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

print("✓ Libraries imported successfully!")
```

## 2. Load Dataset

```python
# Load your employee dataset
# Replace 'employee_dataset_raw.csv' with your file path

try:
    df = pd.read_csv('employee_dataset_raw.csv')
    print(f"✓ Dataset loaded: {len(df)} records, {len(df.columns)} columns")
except FileNotFoundError:
    print("⚠ File not found. Creating sample dataset...")
    
    # Create sample data for demonstration
    np.random.seed(42)
    n = 500
    
    df = pd.DataFrame({
        'EmployeeID': [f'EMP{i:04d}' for i in range(1, n+1)],
        'Age': np.random.randint(22, 65, n),
        'Gender': np.random.choice(['Male', 'Female', 'Other'], n, p=[0.55, 0.42, 0.03]),
        'Department': np.random.choice(['IT', 'HR', 'Finance', 'Marketing', 'Operations', 'Sales'], n),
        'Salary': np.random.randint(50000, 300000, n),
        'YearsOfExperience': np.random.randint(0, 35, n),
        'PerformanceRating': np.random.uniform(1, 5, n).round(2),
        'Tenure': np.random.randint(0, 20, n)
    })
    
    # Add correlation
    df['Salary'] = df['Salary'] + df['YearsOfExperience'] * 5000
    df['Salary'] = df['Salary'].clip(50000, 300000)
    
    print(f"✓ Sample dataset created: {len(df)} records")

# Display first few rows
df.head()
```

## 3. Data Overview

```python
# Basic information
print("Dataset Information:")
print("="*60)
df.info()

print("\n" + "="*60)
print("Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```

## 4. Descriptive Statistics

```python
# Numerical statistics
print("📊 Numerical Variables Summary:")
print("="*60)
df.describe().round(2)
```

```python
# Categorical statistics
print("📋 Categorical Variables Summary:")
print("="*60)

for col in df.select_dtypes(include=['object']).columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

## 5. Visualization Gallery

### 5.1 Salary Analysis by Department

```python
# Average salary by department
avg_salary = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
colors = sns.color_palette('viridis', n_colors=len(avg_salary))
plt.barh(avg_salary.index, avg_salary.values, color=colors, edgecolor='black')

# Add value labels
for i, v in enumerate(avg_salary.values):
    plt.text(v, i, f' {v:,.0f} PKR', va='center', fontsize=11, weight='bold')

plt.title('Average Salary by Department', fontsize=16, weight='bold', pad=20)
plt.xlabel('Average Salary (PKR)', fontsize=12, weight='bold')
plt.ylabel('Department', fontsize=12, weight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.2 Age Distribution

```python
# Age distribution histogram
plt.figure(figsize=(12, 6))
plt.hist(df['Age'], bins=20, color='#3498db', edgecolor='black', alpha=0.7)
plt.axvline(df['Age'].mean(), color='red', linestyle='--', linewidth=2, 
            label=f"Mean: {df['Age'].mean():.1f}")
plt.axvline(df['Age'].median(), color='green', linestyle='--', linewidth=2, 
            label=f"Median: {df['Age'].median():.1f}")

plt.title('Age Distribution Analysis', fontsize=16, weight='bold', pad=20)
plt.xlabel('Age (Years)', fontsize=12, weight='bold')
plt.ylabel('Frequency', fontsize=12, weight='bold')
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.3 Salary Distribution by Department (Box Plot)

```python
# Box plot for salary distribution
plt.figure(figsize=(14, 7))
sns.boxplot(data=df, x='Department', y='Salary', palette='Set2')

plt.title('Salary Distribution by Department', fontsize=16, weight='bold', pad=20)
plt.xlabel('Department', fontsize=12, weight='bold')
plt.ylabel('Salary (PKR)', fontsize=12, weight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.4 Gender Distribution

```python
# Gender pie chart
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(10, 10))
colors = sns.color_palette('pastel', n_colors=len(gender_counts))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%',
        colors=colors, startangle=90, shadow=True)

plt.title('Gender Distribution', fontsize=16, weight='bold', pad=20)
plt.tight_layout()
plt.show()
```

### 5.5 Salary vs Experience Scatter Plot

```python
# Scatter plot with trend line
plt.figure(figsize=(12, 8))

# Plot by department
for dept in df['Department'].unique():
    dept_data = df[df['Department'] == dept]
    plt.scatter(dept_data['YearsOfExperience'], dept_data['Salary'], 
               label=dept, alpha=0.6, s=100, edgecolors='black', linewidth=0.5)

# Add trend line
z = np.polyfit(df['YearsOfExperience'], df['Salary'], 1)
p = np.poly1d(z)
plt.plot(df['YearsOfExperience'].sort_values(), 
         p(df['YearsOfExperience'].sort_values()), 
         "r--", linewidth=2, label='Trend Line')

# Correlation
corr, p_val = stats.pearsonr(df['YearsOfExperience'], df['Salary'])
plt.text(0.05, 0.95, f'Correlation: {corr:.3f}\np-value: {p_val:.4f}',
         transform=plt.gca().transAxes, fontsize=11,
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5),
         verticalalignment='top')

plt.title('Experience vs Salary Relationship', fontsize=16, weight='bold', pad=20)
plt.xlabel('Years of Experience', fontsize=12, weight='bold')
plt.ylabel('Salary (PKR)', fontsize=12, weight='bold')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.6 Correlation Heatmap

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1, mask=mask,
            cbar_kws={"shrink": 0.8}, vmin=-1, vmax=1)

plt.title('Correlation Matrix', fontsize=16, weight='bold', pad=20)
plt.tight_layout()
plt.show()
```

## 6. Statistical Analysis

### 6.1 Correlation Analysis

```python
# Correlation between key variables
print("📊 Correlation Analysis:")
print("="*60)

# Salary vs Experience
corr1, p1 = stats.pearsonr(df['Salary'], df['YearsOfExperience'])
print(f"Salary vs Experience: r = {corr1:.3f}, p = {p1:.4f}")

# Salary vs Age
corr2, p2 = stats.pearsonr(df['Salary'], df['Age'])
print(f"Salary vs Age: r = {corr2:.3f}, p = {p2:.4f}")

# Performance vs Tenure
corr3, p3 = stats.pearsonr(df['PerformanceRating'], df['Tenure'])
print(f"Performance vs Tenure: r = {corr3:.3f}, p = {p3:.4f}")
```

### 6.2 Hypothesis Testing (T-Test)

```python
# Compare salaries between male and female employees
male_salary = df[df['Gender'] == 'Male']['Salary']
female_salary = df[df['Gender'] == 'Female']['Salary']

t_stat, p_value = stats.ttest_ind(male_salary, female_salary)

print("🔬 T-Test: Male vs Female Salaries")
print("="*60)
print(f"Male Mean Salary: {male_salary.mean():,.2f} PKR")
print(f"Female Mean Salary: {female_salary.mean():,.2f} PKR")
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Significant difference (p < 0.05)")
else:
    print("Result: No significant difference (p ≥ 0.05)")
```

### 6.3 Linear Regression

```python
# Linear regression: Experience predicting Salary
from scipy.stats import linregress

slope, intercept, r_value, p_value, std_err = linregress(
    df['YearsOfExperience'], df['Salary']
)

print("📈 Linear Regression: Experience → Salary")
print("="*60)
print(f"Slope: {slope:.2f} PKR/year")
print(f"Intercept: {intercept:.2f} PKR")
print(f"R-squared: {r_value**2:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Standard Error: {std_err:.2f}")
print(f"\nInterpretation: Each additional year of experience")
print(f"is associated with {slope:.2f} PKR increase in salary.")
```

## 7. Key Insights Summary

```python
print("="*60)
print(" "*15 + "KEY INSIGHTS SUMMARY")
print("="*60)

print("\n📊 DATASET OVERVIEW:")
print(f"  • Total Employees: {len(df):,}")
print(f"  • Departments: {df['Department'].nunique()}")
print(f"  • Gender Distribution: {dict(df['Gender'].value_counts())}")

print("\n💰 SALARY INSIGHTS:")
print(f"  • Mean Salary: {df['Salary'].mean():,.2f} PKR")
print(f"  • Median Salary: {df['Salary'].median():,.2f} PKR")
print(f"  • Salary Range: {df['Salary'].min():,.0f} - {df['Salary'].max():,.0f} PKR")

dept_salaries = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
print(f"  • Highest Paying Dept: {dept_salaries.index[0]} ({dept_salaries.values[0]:,.2f} PKR)")
print(f"  • Lowest Paying Dept: {dept_salaries.index[-1]} ({dept_salaries.values[-1]:,.2f} PKR)")

print("\n👥 DEMOGRAPHICS:")
print(f"  • Average Age: {df['Age'].mean():.1f} years")
print(f"  • Average Experience: {df['YearsOfExperience'].mean():.1f} years")
print(f"  • Average Tenure: {df['Tenure'].mean():.1f} years")

print("\n📈 PERFORMANCE:")
print(f"  • Mean Performance Rating: {df['PerformanceRating'].mean():.2f}")
high_perf = (df['PerformanceRating'] >= 4.0).sum()
low_perf = (df['PerformanceRating'] < 3.0).sum()
print(f"  • High Performers (≥4.0): {high_perf} ({high_perf/len(df)*100:.1f}%)")
print(f"  • Need Improvement (<3.0): {low_perf} ({low_perf/len(df)*100:.1f}%)")

print("\n" + "="*60)
```
